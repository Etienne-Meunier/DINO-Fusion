```python
import torch.nn.functional as F 
from einops import rearrange

field = torch.randint(9, [5, 5], dtype=float, requires_grad=True)
#field = torch.rand((5,5), dtype=float, requires_grad=True)

mask = torch.ones_like(field).to(bool)
mask[:2, :5] = False

field_m = torch.where(mask, field, torch.nan)


field_inter = F.interpolate(field_m[None], size=(7), mode='linear')[0]
#field.requires_grad_(True)

l = (field_inter ** 2).nansum()
l.backward()
print(field.grad.sum())
```

```python
import torch.nn.functional as F 
from einops import rearrange

field = torch.randint(9, [5, 5], dtype=float, requires_grad=True)
#field = torch.rand((5,5), dtype=float, requires_grad=True)

mask = torch.ones_like(field).to(bool)
mask[:2, :5] = False

field_m = torch.where(mask, field, torch.nan)


field_inter = F.interpolate(rearrange(field_m[None], 'z x y -> z y x'), size=(7), mode='linear')
field_inter = rearrange(field_inter, 'z y x -> z x y')[0]

l = (field_inter ** 2).nansum()
l.backward()
print(field.grad.sum())
```
